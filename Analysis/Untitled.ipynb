```python

import plotly.express as px
import pandas as pd
import networkx as nx
import numpy as np

# Sample data
data = pd.read_csv("../../GPT/Cluster_plot.csv")


# Ensure Family is treated as a string (in case it contains numeric values)
data['Family'] = data['Family'].astype(str)

# Step 2: Create a layout for each Porphyria_type
positions = {}  # To store the positions for each family

# Expanded offsets for all porphyria types to spread them out
offsets = {
    'AIP': (0, 0),
    'CEP': (3, 0),  # Shifted on x-axis
    'EPP': (0, 3),  # Shifted on y-axis
    'fPCT': (3, 3),  # Shifted on both x and y-axis
    'HCP': (6, 0),  # Further on the x-axis
    'VP': (0, 6),  # Shifted down on the y-axis
    'XLEPP': (6, 3)  # Shifted far on both axes
}

# Step 3: Loop through each Porphyria_type
for porphyria_type, offset in offsets.items():
    subset = data[data['Porphyria_type'] == porphyria_type]

    # Create a graph for this porphyria type
    G = nx.Graph()
    for index, row in subset.iterrows():
        G.add_node(row['Family'], size=row['Pedigree_size'])

    # Calculate force-directed layout
    pos = nx.spring_layout(G, k=0.8, iterations=100)


    # Apply offset to separate the clusters
    for family, coord in pos.items():
        positions[family] = [coord[0] + offset[0], coord[1] + offset[1]]

# Step 4: Extract the x and y positions from the calculated layout
x_pos = []
y_pos = []

# Convert Family to string to match the dictionary keys in `positions`
for family in data['Family'].astype(str):
    if family in positions:
        x_pos.append(positions[family][0])
        y_pos.append(positions[family][1])
    else:
        # Assign default coordinates (e.g., (0, 0)) for missing families
        print(f"Warning: Family {family} is missing from positions.")
        x_pos.append(0)
        y_pos.append(0)

# Add the positions to the dataframe
data['x'] = x_pos
data['y'] = y_pos

# Step 5: Create the packed bubble chart with Plotly
fig = px.scatter(data, x='x', y='y', size='Pedigree_size', color='Porphyria_type',
                 hover_name='Family', size_max=60,
                 title="Porphyria Families: Pedigree Size and Type (Grouped by Porphyria Type)")

# Remove axes and gridlines
fig.update_layout(showlegend=True, xaxis_showgrid=False, yaxis_showgrid=False,
                  xaxis=dict(visible=False), yaxis=dict(visible=False))

# Show the plot
fig.show()


```

```python
import plotly.express as px
import pandas as pd
import numpy as np

# Step 1: Read your data from CSV
data = pd.read_csv("../../GPT/Cluster_plot.csv")

# Ensure Family is treated as a string (in case it contains numeric values)
data['Family'] = data['Family'].astype(str)

# Step 2: Generate hexagonal coordinates
def hexagonal_grid(n):
    """Generate a hexagonal grid with n points."""
    x_coords, y_coords = [], []
    for i in range(n):
        row = i // 10  # Adjust grid width
        col = i % 10
        x = col * np.sqrt(3)
        y = row * 1.5 if col % 2 == 0 else row * 1.5 + 0.75
        x_coords.append(x)
        y_coords.append(y)
    return x_coords, y_coords

# Get hexagonal coordinates for the data
x_hex, y_hex = hexagonal_grid(len(data))
data['x'] = x_hex
data['y'] = y_hex

# Step 3: Create the bubble chart with Plotly using hexagonal grid coordinates
fig = px.scatter(data, x='x', y='y', size='Pedigree_size', color='Porphyria_type',
                 hover_name='Family', size_max=60,
                 title="Porphyria Families: Pedigree Size and Type (Hexagonal Layout)")

# Remove axes and gridlines
fig.update_layout(showlegend=True, xaxis_showgrid=False, yaxis_showgrid=False,
                  xaxis=dict(visible=False), yaxis=dict(visible=False))

# Show the plot
fig.show()
```

```python
import plotly.express as px
import pandas as pd
import networkx as nx
import numpy as np
import plotly.io as pio


# Step 1: Read your data from CSV
data = pd.read_csv("../../GPT/Cluster_plot.csv")

# Ensure Family is treated as a string (in case it contains numeric values)
data['Family'] = data['Family'].astype(str)

# Step 2: Create a graph (G) and add nodes
G = nx.Graph()

# Add nodes (families)
for index, row in data.iterrows():
    G.add_node(row['Family'], size=row['Pedigree_size'])

# Step 3: Add edges based on shared Porphyria_type
# Families with the same Porphyria_type are connected
for porphyria_type in data['Porphyria_type'].unique():
    subset = data[data['Porphyria_type'] == porphyria_type]
    families = subset['Family'].values
    
    # Connect all families with the same Porphyria_type
    for i in range(len(families)):
        for j in range(i+1, len(families)):
            G.add_edge(families[i], families[j])

# Step 4: Use spring_layout to calculate positions
pos = nx.spring_layout(G, k=1, iterations=130)

# Step 5: Define manual offsets for each Porphyria_type
offsets = {
    'AIP': (0, 0),
    'CEP': (1, 0),
    'EPP': (0, 0),
    'fPCT': (0, 0),
    'HCP': (0, 0),
    'VP': (0, 0),
    'XLEPP': (0, 0),
}

# Apply the manual offsets to the positions based on Porphyria_type
for node, position in pos.items():
    porphyria_type = data.loc[data['Family'] == node, 'Porphyria_type'].values[0]
    pos[node][0] += offsets[porphyria_type][0]
    pos[node][1] += offsets[porphyria_type][1]

# Step 5: Extract positions into x and y arrays
x_pos = []
y_pos = []

# Convert Family to string to match the dictionary keys in `pos`
for family in data['Family'].astype(str):
    if family in pos:
        x_pos.append(pos[family][0])
        y_pos.append(pos[family][1])
    else:
        # If a family is missing, assign default coordinates (e.g., (0, 0))
        print(f"Warning: Family {family} is missing from positions.")
        x_pos.append(0)
        y_pos.append(0)

# Add the positions to the dataframe
data['x'] = x_pos
data['y'] = y_pos

# Step 6: Create the bubble chart with Plotly using the calculated positions
fig = px.scatter(data, x='x', y='y', size='Pedigree_size', color='Porphyria_type',
                 hover_name='Family', size_max=60,
                 title="Porphyria Families: Pedigree Size and Type (Grouped by Porphyria Type)")

fig.update_layout(
    width=500,  # Set the desired width
    height=700,  # Set the desired height
    paper_bgcolor='white',  # Set the background to white
    plot_bgcolor='white',   # Set the plot background to white
    showlegend=True, xaxis_showgrid=False, yaxis_showgrid=False,
    xaxis=dict(visible=False), yaxis=dict(visible=False)
)


# Step 7: Export the figure to a high-resolution PNG file with Kaleido
fig.write_image('porphyria_families_high_res.png', format='png', width=500, height=700, scale=3)

# Show the figure in the notebook
fig.show()
```
